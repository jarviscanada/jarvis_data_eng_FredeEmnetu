## Setup


```python
import pyspark.sql.functions as F
bookings = spark.sql("select * from bookings")
facilities = spark.sql("select * from facilities")
members = spark.sql("select * from members")

print(len(facilities.columns))
```

## Question 1

How can you produce a list of the start times for bookings by members named 'David Farrell'?


```python
bookings.join(members, bookings.memid == members.memid, "inner")\
    .filter(members.firstname == "David")\
    .filter(members.surname == "Farrell")\
    .select("starttime")\
    .show()
```

# Question 2

How can you produce a list of the start times for bookings for tennis courts, for the date '2012-09-21'? Return a list of start time and facility name pairings, ordered by the time.  

```python
bookings.join(facilities, bookings.facid == facilities.facid, "inner")\
    .filter(facilities.name.like("Tennis%"))\
    .filter(bookings.starttime >= "2012-09-21")\
    .filter(bookings.starttime < "2012-09-22")\
    .orderBy(bookings.starttime)\
    .select(bookings.starttime, facilities.name)\
    .show()
```

# Question 3

How can you output a list of all members, including the individual who recommended them (if any)? Ensure that results are ordered by (surname, firstname).



```python
members.alias("m").join(
    members.alias("r"),
    F.col("m.recommendedby") == F.col("r.memid"),
    "left"
).select(
    F.col("m.firstname").alias("memfname"),
    F.col("m.surname").alias("memsname"),
    F.col("r.firstname").alias("recfname"),
    F.col("r.surname").alias("recsname")
).orderBy("memsname", "memfname") \
 .show()
```

# Question 4


How can you produce a list of all members who have used a tennis court? Include in your output the name of the court, and the name of the member formatted as a single column. Ensure no duplicate data, and order by the member name followed by the facility name.


```python
members.join(bookings, members.memid == bookings.memid, "inner") \
    .join(facilities, bookings.facid == facilities.facid, "inner") \
    .filter(facilities.name.isin("Tennis Court 2", "Tennis Court 1")) \
    .select(
        F.concat_ws(" ", members.firstname, members.surname).alias("member"),
        facilities.name.alias("facility")
    ) \
    .distinct() \
    .orderBy("member", "facility") \
    .show()
```

# Question 5


How can you output a list of all members, including the individual who recommended them (if any), without using any joins? Ensure that there are no duplicates in the list, and that each firstname + surname pairing is formatted as a column and ordered.


```python

members.alias("m").join(
    members.alias("r"),
    F.col("m.recommendedby") == F.col("r.memid"),
    "left"
).select(
    F.concat_ws(" ", F.col("m.firstname"), F.col("m.surname")).alias("member"),
    F.concat_ws(" ", F.col("r.firstname"), F.col("r.surname")).alias("recommender")
).distinct() \
 .orderBy("member") \
 .show()
```

# Question 6


Produce a count of the number of recommendations each member has made. Order by member ID.


```python
members.filter(F.col("recommendedby").isNotNull()) \
    .groupBy("recommendedby") \
    .agg(F.count("*").alias("count")) \
    .orderBy("recommendedby") \
    .show()
```

# Question 7


Produce a list of the total number of slots booked per facility. For now, just produce an output table consisting of facility id and slots, sorted by facility id.

```python
bookings.groupBy("facid") \
    .agg(F.sum("slots").alias("Total Slots")) \
    .orderBy("facid") \
    .show()
```

# Question 8


Produce a list of the total number of slots booked per facility in the month of September 2012. Produce an output table consisting of facility id and slots, sorted by the number of slots.


```python
bookings.filter(bookings.starttime >= '2012-09-01') \
    .filter(bookings.starttime < '2012-10-01') \
    .groupBy("facid") \
    .agg(F.sum("slots").alias("Total Slots")) \
    .orderBy(F.col("Total Slots")) \
    .select("facid", "Total Slots")\
    .show()
```

# Question 9

Produce a list of the total number of slots booked per facility per month in the year of 2012. Produce an output table consisting of facility id and slots, sorted by the id and month.

```python
bookings.filter(bookings.starttime >= '2012-01-01') \
    .filter(bookings.starttime < '2013-01-01') \
    .withColumn("month", F.month(F.col("starttime"))) \
    .groupBy("facid", "month") \
    .agg(F.sum("slots").alias("sum")) \
    .orderBy("facid", "month") \
    .show()

```

# Question 10

Find the total number of members (including guests) who have made at least one booking.


```python
members.join(bookings, members.memid == bookings.memid, "inner") \
    .filter(bookings.slots >= 1) \
    .select(bookings.memid) \
    .distinct() \
    .agg(F.count(bookings.memid)) \
    .show()
```

# Question 11

Produce a list of each member name, id, and their first booking after September 1st 2012. Order by member ID.


```python
bookings.join(members, members.memid == bookings.memid, "inner") \
    .filter(bookings.starttime >= '2012-09-01') \
    .groupBy(members.memid, members.firstname, members.surname) \
    .agg(F.min(bookings.starttime).alias("firstDate")) \
    .orderBy("memid") \
    .show()
```

# Question 12

Output the names of all members, formatted as 'Surname, Firstname'.

```python
members \
    .withColumn("Full Name", F.concat_ws(", ", members.surname, members.firstname) )\
    .select("Full Name") \
    .show()
```

# Question 13

Perform a case-insensitive search to find all facilities whose name begins with 'tennis'. Retrieve all columns.


```python
facilities \
    .filter(F.upper(facilities.name).like("TENNIS%")) \
    .show()
```

# Question 14

You've noticed that the club's member table has telephone numbers with very inconsistent formatting. You'd like to find all the telephone numbers that contain parentheses, returning the member ID and telephone number sorted by member ID.


```python
members \
    .filter(members.telephone.like("(___)%")) \
    .select("memid", "telephone") \
    .show()
```

# Question 15

You'd like to produce a count of how many members you have whose surname starts with each letter of the alphabet. Sort by the letter, and don't worry about printing out a letter if the count is 0.


```python
members \
    .withColumn("letter", F.substring("surname", 1, 1)) \
    .groupBy("letter") \
    .agg(F.count("letter").alias("count")) \
    .orderBy("letter") \
    .show()
```

# Question 16


Produce a list of all the dates in October 2012. They can be output as a timestamp (with time set to midnight) or a date.


```python
bookings.select(
    F.explode(
        F.sequence(F.to_date(F.lit('2012-10-01')), F.to_date(F.lit('2012-10-31')), F.expr('interval 1 day'))
    ).alias("date")) \
    .show()
```

# Question 17

Return a count of bookings for each month, sorted by month.


```python
bookings \
    .withColumn("yyyy_mm", F.date_format("starttime", "yyyy-MM")) \
    .groupBy("yyyy_mm") \
    .agg(F.count("slots").alias("total_slots")) \
    .orderBy("yyyy_mm") \
    .show()
```
