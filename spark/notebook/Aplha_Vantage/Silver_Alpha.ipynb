# Silver: Cleaning Data

## Grab Tables

```python
aapl = spark.read.table("alpha_vantage.bronze.aapl")
goog = spark.read.table("alpha_vantage.bronze.goog")
inhd = spark.read.table("alpha_vantage.bronze.inhd")
fgi = spark.read.table("alpha_vantage.bronze.fgi")  
```

```python
aapl.head()
aapl.toPandas().shape
aapl.describe()

aapl.dtypes
aapl.toPandas().isnull().sum()
aapl.toPandas().duplicated().sum()
```

```python
# ---- goog ----
goog.head()
goog.toPandas().shape
goog.describe()

goog.dtypes
goog.toPandas().isnull().sum()
goog.toPandas().duplicated().sum()
```

```python
# ---- inhd ----
inhd.head()
inhd.toPandas().shape
inhd.describe()

inhd.dtypes
inhd.toPandas().isnull().sum()
inhd.toPandas().duplicated().sum()
```

```python
# ---- fgi ----
fgi.head()
fgi.toPandas().shape
fgi.describe()

fgi.dtypes
fgi.toPandas().isnull().sum()
fgi.toPandas().duplicated().sum()
```

## Writing tables

```python
aapl.write.mode("overwrite").saveAsTable("alpha_vantage.silver.aapl")
goog.write.mode("overwrite").saveAsTable("alpha_vantage.silver.goog")
inhd.write.mode("overwrite").saveAsTable("alpha_vantage.silver.inhd")
fgi.write.mode("overwrite").saveAsTable("alpha_vantage.silver.fgi")
```
